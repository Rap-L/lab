```python
# Linear regression: fit and plot (robust)
%matplotlib inline
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Load dataset (use workspace path)
df = pd.read_csv('/workspaces/lab/iris.csv')

# Pick two numeric columns (first two) for X and Y if not specified
numeric_cols = df.select_dtypes(include='number').columns.tolist()
if len(numeric_cols) < 2:
    raise ValueError('Need at least two numeric columns to run the example')

x_col = numeric_cols[0]
y_col = numeric_cols[1]

# Prepare X and Y as 2D arrays for sklearn, then flatten for plotting
X = df[[x_col]].values.reshape(-1, 1)
Y = df[[y_col]].values.reshape(-1, 1)

# Train/test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.33, random_state=42)

# Fit linear model
model = LinearRegression()
model.fit(X_train, Y_train)

# Metrics and params
r2 = model.score(X_test, Y_test)
intercept = float(model.intercept_)
coef = float(np.array(model.coef_).reshape(-1)[0])

print(f'Using columns: X="{x_col}", Y="{y_col}"')
print(f'R^2 on test set: {r2:.4f}')
print(f'Intercept: {intercept:.4f}, Coefficient: {coef:.4f}')

# Prepare values for plotting the regression line
x_all = X.reshape(-1)
y_all = Y.reshape(-1)

y_pred_all = intercept + coef * x_all

# Sort by x for a clean line
sort_idx = np.argsort(x_all)
x_sorted = x_all[sort_idx]
y_line = y_pred_all[sort_idx]

plt.figure(figsize=(8,6))
plt.scatter(x_all, y_all, alpha=0.8, label='data')
plt.plot(x_sorted, y_line, 'r-', linewidth=2, label='fit')
plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title('Linear regression: fit vs data')
plt.legend()
plt.show()
```

```python
# Environment & imports
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print('pandas', pd.__version__)
print('seaborn', sns.__version__)

# load dataset
csv_path = 'iris.csv'
df = pd.read_csv(csv_path)

df.head()
```

```python
# Scatter plot (sepal length vs width)

sns.set(style='whitegrid')
plt.figure(figsize=(8,6))

# attempt to normalize column names
def norm(s: str) -> str:
    return s.strip().lower().replace('.', '_').replace(' ', '_')

col_map = {norm(c): c for c in df.columns}

x = col_map.get('sepal_length') or list(df.columns)[0]
y = col_map.get('sepal_width') or (list(df.columns)[1] if len(df.columns) > 1 else list(df.columns)[0])
hue = col_map.get('variety') or col_map.get('species')

if hue in df.columns:
    sns.scatterplot(data=df, x=x, y=y, hue=hue)
else:
    sns.scatterplot(data=df, x=x, y=y)

plt.title('Iris: sepal length vs width')
plt.show()
```
